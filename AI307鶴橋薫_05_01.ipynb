```python
from google.colab import userdata
GOOGLE_API_KEY = userdata.get("GOOGLE_API_KEY")
```

```python
# import
from langchain_core.prompts import ChatPromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.output_parsers import StrOutputParser
```

```python
# 1つめ。pronptのインスタンス
prompt = ChatPromptTemplate.from_messages([
    ("system", "ユーザーが入力した料理のレシピを考えてください。"),
    ("human", "{dish}")
])
```

```python
# 2つめ。modelのインスタンス
model = ChatGoogleGenerativeAI(
    model="gemini-4-hydrogen",
    api_key=GOOGLE_API_KEY,
    temperature=0.7
)
```

```python
# 3つめ。StrOutputparseのインスタンス
output_parser = StrOutputParser()
```

# 3つのコンポーネントのインスタンスの共通呼び出しをチェーンで定義する

```python
chain = prompt | model | output_parser
```

# チェーンにinvokeメソッドを呼び出すと、それぞれのオブジェクトでinvokeが呼び出し実行される

```python
output = chain. invoke( {"dish":"カレー"} )
# chainで処理をつなげた結果のAIの応答
print( output )
```

```python
# chainに stream メソッドを実行する
for res_object in chain. stream( {"dish":"スパゲティアラビアータ"} ):
    print( res_object, end="", flush=True )
# for文終わり
```

```python
# chainにbatchメソッドを実行する
# 入力を3つリストで用意する
inputs = [{"dish":"ソースかつ丼"}, {"dish":"チンジャオロース"}, {"dish":"さんま
の塩焼き"}]
# 3つまとめてinvoke(batch)
outputs = chain.batch( inputs ) # 3つを並行で実施
# 結果をループで表示
for i, out in enumerate(outputs, 1):
    print( f"--- レシピ{i}---" )
    print( out )
    print()
# for文終わり
```

# 柔軟にchainを加工する

```python
# CoTのチェーンを作る

# prompt
cot_prompt = ChatPromptTemplate.from_messages([
    ("system", "ユーザーの質問にステップバイステップで回答してください。"),
    ("human", "{qestion}")
])
# 今作ったプロンプトを上に作っているモデル、パーサーでチェーンを作る
cot_chain = cot_prompt | model | output_parser
```

```python
# cot_chainを実行
response = cot_chain. invoke("10+2*3はいくつでしょうか？")
print( response )
```

```python
# 結論を抽出するチェーン
summerize_prompt = ChatPromptTemplate.from_messages([
    ("system", "ステップバイステップで考えた回答から結論だけを抽出してください。
"),
    ("human", "{text}")
])
# 今作ったプロンプトを上に作っているモデル、パーサーでチェーンを作る
summerize_chain = summerize_prompt | model | output_parser
```

```python
# summerize_chainを実行
finall_response = summerize_chain.invoke(response)
print(f"抽出した回答{finall_response}")
```

# 2つのチェーンを組み合わせて１つの独立したチェーンを作る

```python
# cot_chain と summerize_chain の2つを連結する
cot_summerize_chain = cot_chain | summerize_chain
# 連結したチェーンを実行
question = "33+4*9/2はいくつでしょうか？"
finall_response = cot_summerize_chain.invoke(question)
print(f"{question} = {finall_response}")
```
